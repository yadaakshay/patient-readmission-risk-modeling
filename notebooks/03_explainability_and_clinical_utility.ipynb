### Clinical Interpretation of Top Risk Factors
1. **Prior Inpatient Admissions (`number_inpatient`):** The dominant positive driver of readmission risk. Patients with 1+ previous admissions have more than double the readmission probability.
2. **Discharge Disposition (`discharge_disp_grp`):** Patients discharged with Home Health Care or skilled nursing have significantly higher risk profiles due to underlying frailty.
3. **Glycemic Instability & Monitoring (`A1Cresult`, `insulin`):** Insulin dosage titration and elevated HbA1c (>8%) represent acute metabolic volatility.
4. **Healthcare Intensity & Polypharmacy (`meds_per_day`, `time_in_hospital`):** Extended stay duration combined with complex medication regimens flags patients with multimorbidity.


### Actionable Clinical Protocol:
| Risk Tier | Risk Threshold | Care Management Protocol |
| :--- | :---: | :--- |
| **Low Risk** | $< 6.35\%$ | Routine discharge summary, primary care follow-up within 14–30 days, portal medication list review. |
| **Moderate Risk** | $6.35\% - 11.44\%$ | Targeted disease education, follow-up clinic appointment booked within 7 days, telephone nurse check-in at 48 hours. |
| **High Risk** | $> 11.44\%$ | **Intensive Transition of Care:** Nurse navigator assigned, bedside medication delivery, 72-hour clinical visit, and pharmacist medication reconciliation. |
